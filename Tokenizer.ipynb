```
from transformers import AutoTokenizer
```

## Word Piece

```
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
```

```
tokenizer.tokenize("unhappily")
```

```
tokenizer.tokenize("lowercaseable")
```

```
tokenizer.tokenize(["Low", "Lower", "Lowest"])
```

## Byte paired Encoding (BPE)

```
tokenizer = AutoTokenizer.from_pretrained("gpt2")
```

```
tokenizer.tokenize("unhappily")
```

```
tokenizer.tokenize("lowercaseable")
```

```
tokenizer.tokenize(["Low", "Lower", "Lowest"])
```

## Unigram or Sentence piece

```
tokenizer = AutoTokenizer.from_pretrained("t5-base")
```

```
tokenizer.tokenize("unhappily")
```

```
tokenizer.tokenize("lowercaseable")
```

```
tokenizer.tokenize(["Low", "Lower", "Lowest"])
```

# Special Token

```
text_1 = "I love You"
text_2 = "I am happy"

encode = tokenizer(text_1, text_2)

tokenizer.convert_ids_to_tokens(encode["input_ids"])
```

```
encode = tokenizer(text_1)

tokenizer.convert_ids_to_tokens(encode["input_ids"])
```

### Padding

```
text_a = "I am AI Engineer"
text_b = "Currently learn a LLM Concepts for my career"

encode_1 = tokenizer(text_a)
encode_2 = tokenizer(text_b)

print(encode_1["input_ids"])
print(encode_2["input_ids"])
```

```
encode = tokenizer([text_a, text_b], padding = True, return_tensors="pt")

encode["input_ids"]
```

```
for token in tokenizer.all_special_tokens:
    print(token, "->", tokenizer.convert_tokens_to_ids(token))
```
